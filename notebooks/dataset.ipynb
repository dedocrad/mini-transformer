```python
import random
import string
import pandas as pd
```

```python
seed = 42
random.seed(seed)
```

```python
def generate_random_string():
    letters = string.ascii_letters
    random_string = ''.join(random.choice(letters) for _ in range(random.randint(1,20)))
    return random_string
```

```python
n = 100000
data = []

while len(data) < n:
    word = generate_random_string()
    data.append({'word' : word, 'rev_word' : word[::-1]})

df = pd.DataFrame(data)
```
